```python
import zipfile

zip_path = "/content/drive/MyDrive/NFL_GNN_Project/data (1)/nfl-big-data-bowl-2026-prediction.zip"

with zipfile.ZipFile(zip_path, 'r') as z:
    names = z.namelist()

print("zip 内前 50 个文件：")
for n in names[:50]:
    print(n)

print("\n包含 train_input 的文件：", [n for n in names if "train_input" in n.lower()])
print("包含 train_output 的文件：", [n for n in names if "train_output" in n.lower()])
print("包含 csv 的文件前 30 个：", [n for n in names if n.lower().endswith(".csv")][:30])
```

```python
from nfl_bdb_preprocess_to_csv import export_preprocessed_csvs

manifest = export_preprocessed_csvs(
    data_dir=DATA_DIR,
    output_dir=PREPROC_DIR,
    include_test=True,
    output_prefix="preprocessed",
    save_manifest=True,
)

print("Preprocessing done.")
print(manifest)
```
